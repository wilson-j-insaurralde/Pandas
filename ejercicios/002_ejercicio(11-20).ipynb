Exercise 14:

Concatenate two DataFrames.

```python
import pandas as pd
data1={'x':[1,2,3,4]}
data2={'y':[5,6,7,8]}
df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)
unir_df=pd.concat([df1,df2],axis=1)
print(unir_df)
```

Exercise 15:

Merge two DataFrames on a key.

```python
import pandas as pd 
data1 = {'key': ['X', 'Y', 'Z'], 'value1': [1, 2, 3]}
data2 = {'key': ['X', 'Y', 'D'], 'value2': [4, 5, 6]}
df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)
unir_df=pd.merge(df1,df2,on='key',how='outer')
print(unir_df)
```

Exercise 16:

Create a pivot table from a DataFrame.

```python
import pandas as pd
data = {'X': ['foo', 'foo', 'bar', 'bar'], 'Y': ['one', 'two', 'one', 'two'], 'Z': [1, 2, 3, 4]}
df = pd.DataFrame(data)
pivot_table = df.pivot_table(values='Z', index='X', columns='Y')
print(pivot_table)
```

Exercise 17:

Reshape a DataFrame from long to wide format.

```python
import pandas as pd
data = {'X': ['foo', 'foo', 'bar', 'bar'], 'Y': ['one', 'two', 'one', 'two'], 'Z': [1, 2, 3, 4]}
df = pd.DataFrame(data)
ancho_df=df.pivot(index='X', columns='Y', values='Z')
print(df)
print(ancho_df)
```

Exercise 18:

Calculate the correlation between columns in a DataFrame.

```python
import pandas as pd
data = {'X': [1, 2, 3, 4], 'Y': [4, 3, 2, 1]}
df = pd.DataFrame(data)
correlaccion = df.corr()
print(df)
print(correlaccion)
```

Exercise 19:

Iterate over rows in a DataFrame using iterrows().

```python
import pandas as pd
data = {'X': [1, 2, 3], 'Y': [4, 5, 6]}
df = pd.DataFrame(data)
for index, row in df.iterrows():
    print(index, row['X'], row['Y'])
```

Exercise 20:

Apply a function to each element in a DataFrame.

```python
import pandas as pd 
data = {'X': [1, 2, 3], 'Y': [4, 5, 6]}
df1 = pd.DataFrame(data)
df2 = df1.map(lambda x: x * 2)
print(df1)
print(df2)
```
