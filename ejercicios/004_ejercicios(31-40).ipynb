Exercise 32:

Calculate the rank of values in a DataFrame.

```python
import pandas as pd
data = {'X': [3, 1, 4, 1], 'Y': [2, 3, 1, 4]}
df = pd.DataFrame(data)
df['Rank'] = df['X'].rank()
print(df)
```

Exercise 33:

Change the data type of a column.

```python
import pandas as pd 
data = {'X': ['1', '2', '3']}
df= pd.DataFrame(data)
df['X']=df['X'].astype(int)

print (df)
```

Exercise 34:

Filter rows based on string matching.

```python
import pandas as pd 
data = {'X': ['foo', 'bar', 'baz', 'qux']}
df = pd.DataFrame(data)
filtro_df=df[df['X'].str.contains('ba')]
print(filtro_df)
```

Exercise 35:

Create a DataFrame with specified row and column labels.

```python
import pandas as pd
data = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
df = pd.DataFrame(data, index=['row1', 'row2', 'row3'], columns=['col1', 'col2', 'col3'])
print(df)
```

Exercise 36:

Transpose a DataFrame.

```python
import pandas as pd
data = {'X': [1, 2, 3], 'Y': [4, 5, 6]}
df = pd.DataFrame(data)
transposed_df = df.T
print(transposed_df)
```

Exercise 37:

Set a column as the index of a DataFrame.

```python
import pandas as pd
data = {'X': [1, 2, 3], 'Y': [4, 5, 6]}
df = pd.DataFrame(data)
print(df)
df.set_index('X', inplace=True)
print(df)
```

Exercise 38:

Reset the index of a DataFrame.

```python
import pandas as pd
data = {'X': [1, 2, 3], 'Y': [4, 5, 6]}
df = pd.DataFrame(data)
df.set_index('X', inplace=True)
df.reset_index(inplace=True)
print(df)
```

Exercise 39:

Add a prefix or suffix to column names.

```python
import pandas as pd
data = {'X': [1, 2, 3], 'Y': [4, 5, 6]}
df = pd.DataFrame(data)
df = df.add_prefix('col_')
print(df)
```

Exercise 40:

Filter rows based on datetime index.

```python
import pandas as pd
date_range = pd.date_range(start='1/1/2020', periods=5, freq='D')
data = {'X': [1, 2, 3, 4, 5]}
df = pd.DataFrame(data, index=date_range)
filtered_df = df['2020-01-03':'2020-01-05']
print(filtered_df)
```
