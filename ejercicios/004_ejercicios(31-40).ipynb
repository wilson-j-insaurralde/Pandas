Exercise 32:

Calculate the rank of values in a DataFrame.

```python
import pandas as pd
data = {'X': [3, 1, 4, 1], 'Y': [2, 3, 1, 4]}
df = pd.DataFrame(data)
df['Rank'] = df['X'].rank()
print(df)
```

Exercise 33:

Change the data type of a column.

```python
import pandas as pd 
data = {'X': ['1', '2', '3']}
df= pd.DataFrame(data)
df['X']=df['X'].astype(int)

print (df)
```

Exercise 34:

Filter rows based on string matching.

```python
import pandas as pd 
data = {'X': ['foo', 'bar', 'baz', 'qux']}
df = pd.DataFrame(data)
filtro_df=df[df['X'].str.contains('ba')]
print(filtro_df)
```
