Create a DataFrame from a list of dictionaries.

Exercise 22:

Rename columns in a DataFrame.

```python
import pandas as pd
data={'X': [1, 2, 3], 'Y': [4, 5, 6]}
df = pd.DataFrame(data)
print(df)

df.rename(columns={'X': 'z', 'Y': 'k'}, inplace=True)
print(df)
```

Exercise 23:

Filter rows by multiple conditions.

```python
import pandas as pd 
data={'X': [1, 2, 3, 4], 'Y': [4, 5, 6, 7]}
df = pd.DataFrame(data)
filtrar_df=df[(df['X']>2)&(df['Y']<7)]

print(filtrar_df)
```

Exercise 24:

Calculate the cumulative sum of a column.

```python
import pandas as pd 
data={'X':[1,2,3,4]}
df=pd.DataFrame(data)
df['suma-acumulativa']=df['X'].cumsum()
print(df)
```

Exercise 25:

Drop rows with missing values.

```python
import pandas as pd 
data={'X': [1, 2, None, 4], 'Y': [4, 5, 6, None]}
df=pd.DataFrame(data)
print(df)
print("----"*8)
df.dropna(inplace=True)
print (df)
```

Exercise 26:

Replace values in a DataFrame based on a condition.

```python
import pandas as pd 
data = {'X': [1, 2, 3, 4], 'Y': [5, 6, 7, 8]}
df=pd.DataFrame(data)
df.loc[df['X']>2,'Y']=0
print(df)
```

Exercise 27:

Create a DataFrame with a MultiIndex.

```python
import pandas as pd 
arrays = [['X', 'X', 'Y', 'Y'], [1, 2, 1, 2]]
index = pd.MultiIndex.from_arrays(arrays, names=('Group', 'Number'))
data = {'Value': [10, 20, 30, 40]}
df = pd.DataFrame(data, index=index)
print(df)
```

Exercise 28:

Calculate the rolling mean of a column.

```python
import pandas as pd
data = {'X': [1, 2, 3, 4, 5, 6]}
df = pd.DataFrame(data)
df['media-movil'] = df['X'].rolling(window=3).mean()
print(df)
```

Exercise 29:

Create a DataFrame from a list of tuples.

Exercise 30:

Add a row to a DataFrame.

```python
import pandas as pd
data = {'X': [1, 2], 'Y': [3, 4]}
df = pd.DataFrame(data)
new_row = pd.DataFrame({'X': [5], 'Y': [6]})
df = pd.concat([df, new_row], ignore_index=True)
print(df)
```
