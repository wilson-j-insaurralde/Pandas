Cuando trabajás con conjuntos grandes de datos, resumir información es vital. Con Pandas podés agrupar filas y aplicar cálculos agregados que condensan patrones clave.

La combinación de DataFrame y el método groupby() abre la puerta a estadísticas resumidas en segundos.

9.1 Uso de groupby
groupby separa un DataFrame en grupos según columnas elegidas y prepara los datos para agregaciones.

Ejemplo de DataFrame

```python
import pandas as pd
datos = {
    "Empleado": ["Ana", "Luis", "Carla", "Pedro", "María", "Jorge"],
    "Departamento": ["Ventas", "Ventas", "IT", "IT", "Ventas", "IT"],
    "Salario": [55000, 72000, 61000, 83000, 59000, 75000],
    "Edad": [23, 35, 29, 42, 31, 38]
}
df = pd.DataFrame(datos)
print(df)
```

Agrupar por una columna

```python
grupo = df.groupby("Departamento")
print(grupo)
```

👉 El resultado es un objeto DataFrameGroupBy; la agregación llega cuando aplicás funciones como mean() o agg().

9.2 Funciones de agregación (sum, mean, count)
Una vez agrupados, aplicás funciones de agregación para obtener indicadores por grupo.

Calcular la media por grupo

```python
print(df.groupby("Departamento")["Salario"].mean())
```

Calcular varias métricas

```python
print(df.groupby("Departamento")["Edad"].sum())
print(df.groupby("Departamento")["Edad"].count())
```

Agregación sobre columnas numéricas

```python
df_promedios = df.groupby("Departamento")[["Salario", "Edad"]].mean()
print(df_promedios)
```

👉 Seleccioná las columnas numéricas con doble corchete (o usá numeric_only=True) para evitar errores en versiones recientes de Pandas.

9.3 Agregaciones múltiples y personalizadas
Con agg() podés combinar funciones estándar y lógicas propias en una sola llamada.

Agregaciones múltiples

```python
print(df.groupby("Departamento")["Salario"].agg(["sum", "mean", "max"]))
```

Agregación sobre varias columnas

```python
print(df.groupby("Departamento").agg({
    "Salario": ["mean", "max"],
    "Edad": ["min", "max"]
}))
```

Funciones personalizadas con agg()

```python
def rango(x):
    return x.max() - x.min()
print(df.groupby("Departamento").agg({
    "Salario": rango,
    "Edad": rango
}))
```

👉 Con funciones personalizadas obtenés indicadores propios, como el rango de salarios o edades por equipo.

Resumen del tema 9
groupby() agrupa datos según una o varias columnas.
Las funciones estándar (sum, mean, count, max, min) generan resúmenes rápidos.
agg() combina múltiples funciones y admite lógicas personalizadas.
Las agregaciones ayudan a crear reportes comparativos en pocas líneas de código.
