En casi todos los datasets reales aparecen valores faltantes: celdas vacías, datos perdidos o inconsistentes. Pandas ofrece funciones muy completas para detectarlos, eliminarlos o reemplazarlos sin perder contexto analítico.

Trabajaremos con un DataFrame de ejemplo y aprovecharemos la integración con NumPy para crear valores nulos representados como NaN (Not a Number).

6.1 Detección de valores nulos (isnull, notnull)
En Pandas los valores faltantes se representan como NaN. Podemos identificarlos rápidamente usando los métodos isnull() y notnull().

Crear un DataFrame con valores nulos

```python
import pandas as pd
import numpy as np

datos = {
    "Nombre": ["Ana", "Luis", "Carla", "Pedro", "María"],
    "Edad": [23, np.nan, 29, 42, None],
    "Ciudad": ["Córdoba", "Buenos Aires", None, "Mendoza", "Salta"],
    "Salario": [55000, 72000, None, 83000, 59000]
}

df = pd.DataFrame(datos)
print(df)
```

isnull() detecta valores nulos

```python
print(df.isnull())
```

notnull() detecta valores NO nulos

```python
print(df.notnull())
```

Contar valores nulos por columna

```python
print(df.isnull().sum())
```

👉 Con esto sabés en qué columnas y cuántos datos faltan.

6.2 Eliminación de datos (dropna)
A veces lo más simple es eliminar filas o columnas que contienen valores faltantes. dropna() ofrece varios parámetros para controlar la limpieza.

```python
# Eliminar filas con cualquier valor nulo
df_sin_nulos = df.dropna()
print(df_sin_nulos)

# Eliminar columnas con cualquier valor nulo
df_sin_columnas = df.dropna(axis=1)
print(df_sin_columnas)

# Eliminar filas solo si todos los valores son nulos
df_sin_todos = df.dropna(how="all")

# Eliminar filas si tienen menos de un mínimo de valores no nulos
df_min = df.dropna(thresh=3)  # conserva filas con al menos 3 valores válidos
```

👉 Esta estrategia puede ser útil, pero cuidado: eliminar datos reduce el tamaño del dataset.

```python
6.3 Imputación y reemplazo (fillna, replace)
Cuando no queremos eliminar información, podemos rellenar valores faltantes con distintos métodos o reemplazar datos problemáticos.

Rellenar con un valor fijo
```

```python
df_relleno = df.fillna(0)
print(df_relleno)
```

Rellenar con la media, mediana o moda

```python
# Rellenar Edad con la media
#df["Edad"].fillna(df["Edad"].mean(), inplace=True)

# Rellenar Salario con la mediana
#df["Salario"].fillna(df["Salario"].median(), inplace=True)

# Rellenar Ciudad con un valor por defecto
#df["Ciudad"].fillna("Desconocida", inplace=True)

# Rellenar Edad con la media (Forma moderna sin inplace)
df["Edad"] = df["Edad"].fillna(df["Edad"].mean())

# Rellenar Salario con la mediana (Forma moderna sin inplace)
df["Salario"] = df["Salario"].fillna(df["Salario"].median())

# Rellenar Ciudad con un valor por defecto (Forma moderna sin inplace)
df["Ciudad"] = df["Ciudad"].fillna("Desconocida")
```

Rellenar hacia adelante o hacia atrás

```python
# Forward fill (usa el valor anterior)
#df_ffill = df.fillna(method="ffill")

# Backward fill (usa el valor siguiente)
#df_bfill = df.fillna(method="bfill")

# Forward fill (usa el valor anterior)
df_ffill = df.ffill()

# Backward fill (usa el valor siguiente)
df_bfill = df.bfill()
```

Reemplazar valores específicos (replace)

```python
# Reemplazar None por un texto fijo
#df["Ciudad"].replace({None: "Sin ciudad"}, inplace=True)

# Reemplazar un salario por otro valor
#df["Salario"].replace({59000: 60000}, inplace=True)


# Reemplazar None por un texto fijo (Forma moderna sin inplace)
df["Ciudad"] = df["Ciudad"].replace({None: "Sin ciudad"})

# Reemplazar un salario por otro valor (Forma moderna sin inplace)
df["Salario"] = df["Salario"].replace({59000: 60000})
```

👉 El manejo correcto de valores faltantes es esencial: un mal tratamiento puede sesgar resultados o hacer fallar modelos.

Resumen del tema 6
isnull(), notnull() y .sum() sirven para detectar y cuantificar valores faltantes.
dropna() elimina filas o columnas según el criterio de nulos.
fillna() rellena con valores fijos, media, mediana o estrategias forward/backward fill.
replace() reemplaza valores específicos, incluyendo None o outliers puntuales.
